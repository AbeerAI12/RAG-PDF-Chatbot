```python
from google.colab import userdata
from google import genai

GOOGLE_API_KEY = userdata.get('GEMINI_API_KEY')  # غيرنا الاسم هون بس
client = genai.Client(api_key=GOOGLE_API_KEY)

print("تم الاتصال بنجاح ✅")
```

```python
from pypdf import PdfReader

reader = PdfReader("Course Outline -Machine Learning-python.pdf")

print(f"عدد الصفحات: {len(reader.pages)}")
print("---")
print(reader.pages[0].extract_text())
```

```python
# اقرأ كل الصفحات وضمهم بنص واحد
full_text = ""

for i, page in enumerate(reader.pages):
    text = page.extract_text()
    full_text += text + "\n"
    print(f"صفحة {i+1}: {len(text)} حرف")

print("---")
print(f"إجمالي عدد الأحرف بكل الملف: {len(full_text)}")
```

```python
def split_text(text, chunk_size=500, overlap=50):
    chunks = []
    start = 0

    while start < len(text):
        end = start + chunk_size
        chunk = text[start:end]
        chunks.append(chunk)
        start += chunk_size - overlap  # نرجع للخلف شوي عشان التداخل

    return chunks

# طبق الدالة على النص الكامل
chunks = split_text(full_text)

print(f"عدد القطع: {len(chunks)}")
print("---")
print("أول قطعة كمثال:")
print(chunks[0])
```

```python
result = client.models.embed_content(
    model="gemini-embedding-001",
    contents=chunks[0]
)

embedding = result.embeddings[0].values

print(f"طول المتجه (عدد الأرقام): {len(embedding)}")
print("---")
print("أول 10 أرقام من المتجه كمثال:")
print(embedding[:10])
```

```python
# نحول كل القطع لـ embeddings
result_all = client.models.embed_content(
    model="gemini-embedding-001",
    contents=chunks  # هلق منبعت القائمة كاملة، مش قطعة وحدة بس
)

all_embeddings = [emb.values for emb in result_all.embeddings]

print(f"عدد الـ embeddings يلي طلعوا: {len(all_embeddings)}")
print(f"طول كل واحد فيهم: {len(all_embeddings[0])}")
```

```python
import chromadb

# أنشئ قاعدة بيانات مؤقتة بالذاكرة (كافية لمشروعنا التعليمي)
chroma_client = chromadb.Client()

# أنشئ "collection" - يعني جدول/مجموعة نخزن فيها القطع
collection = chroma_client.create_collection(name="course_chunks")

# خزّن كل قطعة مع الـ embedding يلي يمثلها
collection.add(
    documents=chunks,                                    # النص الأصلي لكل قطعة
    embeddings=all_embeddings,                            # المتجه الرقمي لكل قطعة
    ids=[f"chunk_{i}" for i in range(len(chunks))]        # معرف فريد لكل قطعة
)

print(f"تم تخزين {collection.count()} قطعة بنجاح ✅")
```

```python
# اكتب سؤال تجريبي (لازم يكون متعلق بمحتوى الملف يلي رفعته)
query = "شو مدة كورس Python؟"

# حول السؤال لـ embedding (بنفس الطريقة يلي حولنا فيها القطع)
query_embedding = client.models.embed_content(
    model="gemini-embedding-001",
    contents=query
).embeddings[0].values

# دور بالـ ChromaDB عن أقرب 3 قطع لهاد السؤال
results = collection.query(
    query_embeddings=[query_embedding],
    n_results=3
)

# اطبع النتائج
for i, doc in enumerate(results['documents'][0]):
    print(f"القطعة {i+1}:")
    print(doc)
    print("---")
```

```python
# The ask_rag function definition is moved to the Gradio cell to ensure scope.
```

```python
import gradio as gr

# دالة للربط مع واجهة Gradio
def chat_with_pdf(question):
    # تعريف ask_rag كدالة داخلية لضمان نطاقها داخل chat_with_pdf
    def ask_rag(query, n_results=3):
        # 1. حول السؤال لـ embedding
        query_embedding = client.models.embed_content(
            model="gemini-embedding-001",
            contents=query
        ).embeddings[0].values

        # 2. دور عن أقرب القطع
        results = collection.query(
            query_embeddings=[query_embedding],
            n_results=n_results
        )

        # 3. اجمع القطع بنص واحد (context)
        context = "\n\n".join(results['documents'][0])

        # 4. ابني الـ prompt وابعته لـ Gemini
        prompt = f"""أجب على السؤال بناءً على السياق التالي فقط. إذا كانت المعلومات غير كافية، قل ذلك بوضوح.\n\nالسياق:\n{context}\n\nالسؤال: {query}\n\nالإجابة:"""

        response = client.models.generate_content(
            model="gemini-2.5-flash",
            contents=prompt
        )

        return response.text

    if not question.strip():
        return "الرجاء إدخال سؤال."
    # استدعاء الدالة الداخلية
    return ask_rag(question)

# تصميم واجهة الويب
demo = gr.Interface(
    fn=chat_with_pdf,
    inputs=gr.Textbox(lines=2, placeholder="اكتبي سؤالك هنا عن محتوى الملف... (مثلاً: ما هي مدة الكورس؟)"),
    outputs=gr.Textbox(label="إجابة الذكاء الاصطناعي (RAG)"),
    title="تطبيق محادثة الملفات الذكي 🤖",
    description="هذا النظام يقرأ ملف الـ PDF الخاص بك ويجيب بدقة بناءً على محتواه فقط."
)

# تشغيل الواجهة (سيظهر لك رابط مباشر)
demo.launch(share=True, debug=True)
```
